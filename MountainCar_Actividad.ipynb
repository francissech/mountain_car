## Resultados obtenidos

Se implementaron y entrenaron dos agentes de aprendizaje por refuerzo
para resolver el entorno MountainCar-v0: Q-Learning tabular y Deep
Q-Network (DQN).

El agente Q-Learning utilizó una discretización del espacio continuo
de estados, formado por la posición y velocidad del automóvil.
Después del entrenamiento alcanzó una recompensa media de
-128.30 ± 21.69 durante 10 episodios de evaluación y consiguió
alcanzar la bandera en 10 de 10 episodios.

El agente DQN sustituyó la tabla Q por una red neuronal completamente
conectada. La red recibió como entrada la posición y velocidad del
automóvil y produjo tres valores Q correspondientes a las tres
acciones disponibles.

Durante los primeros experimentos se observó que el DQN con una
exploración epsilon-greedy convencional presentaba dificultades para
aprender, obteniendo recompensas próximas a -200.

Esto ocurre porque MountainCar requiere que el automóvil mantenga
acciones durante varios pasos para acumular impulso y poder ascender
la montaña.

Por este motivo se implementó una estrategia de exploración
temporalmente correlacionada, manteniendo una acción exploratoria
durante 20 pasos. Después de esta modificación el aprendizaje mejoró
considerablemente.

En la evaluación final, el DQN obtuvo una recompensa media de
-103.20 ± 7.19 y alcanzó la bandera en 10 de 10 episodios.

## Comparación de los métodos

### Q-Learning

Q-Learning almacena explícitamente los valores Q(s,a) en una tabla.
Debido a que MountainCar posee un espacio de estados continuo, fue
necesario discretizar las variables posición y velocidad.

Ventajas:
- Implementación relativamente sencilla.
- Bajo costo computacional.
- Fácil interpretación de los valores Q.
- Convergencia estable para este problema discretizado.

Limitaciones:
- Requiere discretizar espacios continuos.
- La tabla Q puede crecer rápidamente cuando aumenta el número de
  variables de estado.
- Presenta dificultades para escalar a problemas de alta dimensión.

### DQN

DQN aproxima la función Q(s,a) mediante una red neuronal, evitando
la necesidad de almacenar explícitamente una tabla para cada estado.

Ventajas:
- Puede trabajar directamente con estados continuos.
- Permite generalizar entre estados similares.
- Es más adecuado para problemas de mayor dimensionalidad.

Limitaciones:
- Mayor complejidad de implementación.
- Mayor costo computacional.
- El entrenamiento puede ser inestable.
- Es sensible a hiperparámetros y a la estrategia de exploración.

En este experimento ambos agentes alcanzaron la bandera en el 100 %
de los episodios de evaluación. El DQN presentó una recompensa media
de -103.20 ± 7.19, mientras que Q-Learning obtuvo -128.30 ± 21.69.
Por tanto, en estas evaluaciones el DQN necesitó, en promedio, menos
pasos para completar la tarea y presentó menor variabilidad entre
episodios.

# Conclusiones

En esta actividad se implementaron dos enfoques de aprendizaje por
refuerzo para resolver MountainCar-v0.

Q-Learning permitió comprender de forma directa la actualización de
los valores acción mediante la ecuación de Bellman. Sin embargo, fue
necesario discretizar el espacio continuo de estados.

DQN reemplazó la tabla Q por una red neuronal capaz de aproximar la
función Q. Su implementación fue más compleja debido al uso de una
red neuronal, experience replay y una target network.

Un resultado importante fue comprobar que una exploración
epsilon-greedy completamente aleatoria no resultó suficiente para
obtener un aprendizaje estable en este problema. MountainCar requiere
acciones sostenidas que permitan acumular impulso. La incorporación
de una exploración temporalmente correlacionada permitió mejorar
considerablemente el entrenamiento.

En la evaluación final ambos agentes alcanzaron la meta en 10 de 10
episodios. Q-Learning obtuvo una recompensa media de
-128.30 ± 21.69 y DQN obtuvo -103.20 ± 7.19.

Los resultados muestran que, bajo la configuración utilizada en este
experimento, DQN completó la tarea en menos pasos en promedio y con
menor variabilidad, aunque requirió una implementación y un proceso
de entrenamiento más complejos.
